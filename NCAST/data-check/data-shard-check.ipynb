# Original

```python
original_example = torch.load("/work/scratch-nopw2/mendrika/OB/ncast/raw/shards/t3/train_t3/shard_001.pt")
```

```python
Y = original_example["Y"]
X = original_example["X"]
```

```python
scaler = torch.load("/home/users/mendrika/NCAST/NCAST/scaler/scaler_realcores.pt", weights_only=False)

mean = np.asarray(scaler["mean"])
scale = np.asarray(scaler["scale"])

MASK_COL_INDEX = 12
COLS_TO_SCALE = range(4, 12)
```

```python
flat = X.numpy().reshape(-1, X.shape[-1])

flat[:, COLS_TO_SCALE] = (flat[:, COLS_TO_SCALE] - mean) / scale

X_scaled = flat.reshape(X.shape)
X_scaled = torch.tensor(X_scaled, dtype=torch.float32)
```
